### Puzzle: Reverse a List
Refactor the following function to use Python slicing instead of a loop.

````python
def reverse_list(lst):
    result = []
    for i in range(len(lst)-1, -1, -1):
        result.append(lst[i])
    return result

# Example
print(reverse_list([1, 2, 3, 4]))  # Expected output: [4, 3, 2, 1]
````
