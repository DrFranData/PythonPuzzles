### Puzzle: Count Character Frequencies
The function below creates a dictionary that maps each character in a string to the number of times it appears. Refactor it using dictionary comprehension.

````python
def char_frequency(text):
    freq = {}
    for char in set(text):
        freq[char] = text.count(char)
    return freq

# Example
print(char_frequency("banana"))  # Expected output: {'b': 1, 'a': 3, 'n': 2}
````

```python
def char_frequency(text):
    return {char: text.count(char) for char in set(text)}
```
