### Puzzle: Optimize List Comprehension
Refactor the function using list comprehension.

````python
def square_list(n):
    result = []
    for i in range(n+1):
        result.append(i*i)
    return result
````

````python
print(square_list(3)) # expected output: [0, 1, 4, 9]
````

<details>
  <summary><b>Show Solution</b></summary>
```python
def square_list(n):
    return [i*i for i in range(n+1)]
```
