##🧩 Puzzle 6: "Sum Those Numbers!"
###📖 Concepts You Need:
**Lists**: Storing multiple numbers in a list.

**Indexing**: Accessing specific elements in a list.

**Summation**: Using `sum(` to add up numbers.
###💡 Puzzle:
Write a Python program that asks the user for three numbers, stores them in a list, and prints their total sum.

###🖥 Example:
```
Enter first number: 4
Enter second number: 7
Enter third number: 2
The total sum is 13.
```
###🔑 Hints:

Use a list to store the numbers: `[num1, num2, num3]`.

Use `sum(my_list)` to add them up.

````python
# your code here
````

<details>
  <summary><b>✅ Show Solution</b></summary>

```python
# Ask the user for three numbers and store them in a list
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

numbers = [num1, num2, num3]

# Calculate the sum
total = sum(numbers)

# Print the result
print("The total sum is " + str(total) + ".")
```
