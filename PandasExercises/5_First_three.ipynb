#🧩 Puzzle 5: "First Three, Please!"
###📖 Concepts You Need:
**Slicing**: Extracting part of a string using `string[start:end]`.
###💡 Puzzle:
Write a Python program that asks the user for a word and prints only the first three letters.

###🖥 Example:
```
Enter a word:
Elephant
The first three letters are: Ele
```

````python
# Your code here
````

<details>
  <summary><b>✅ Show Solution</b></summary>

```python
# Ask the user for a word
word = input("Enter a word: ")

# Get the first three letters using slicing
first_three = word[:3]

# Print the result
print("The first three letters are: " + first_three)

```
