###💡 Puzzle:
Write a Python program that asks the user for four numbers, stores them in a list, calculates the average, and prints the result.

###🖥 Example:
```
Enter four numbers:
10
20
30
40
The average is 25.0.
```

```python
# Ask the user for four numbers and store them in a list
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))
num4 = int(input("Enter fourth number: "))

numbers = [num1, num2, num3, num4]

# Calculate the average
average = sum(numbers) / len(numbers)

# Print the result
print("The average is " + str(average) + ".")

```
